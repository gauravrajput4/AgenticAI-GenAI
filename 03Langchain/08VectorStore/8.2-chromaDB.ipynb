### ChromaDB
Chroma is a AI-native open-source vector database focused on developer productivity and happiness. Chroma is licensed under Apache 2.0

```python
# Building a sample vectorDB
from langchain_chroma import Chroma
from langchain_community.document_loaders import TextLoader
from langchain_community.embeddings import  OllamaEmbeddings
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
loader=TextLoader('Speech.txt')
data=loader.load()
data
```

```python
text_splitter=RecursiveCharacterTextSplitter(
    chunk_size=100,
    chunk_overlap=50
)
splits=text_splitter.split_documents(data)
```

```python
embeddings=OllamaEmbeddings(model="gemma2:2b")
vectorDB=Chroma.from_documents(documents=splits,embedding=embeddings)
vectorDB
```

```python
# Querying the vectorDB
query="What is the speech about?"
docs=vectorDB.similarity_search(query)
docs[0].page_content
```

```python
# Saving and Loading Chroma Vectorstore
vectorDB=Chroma.from_documents(documents=splits,embedding=embeddings,persist_directory="./chroma_index")
```

```python
# load from disk
db2=Chroma(persist_directory="./chroma_index",embedding_function=embeddings)
```

```python
# Querying the vectorDB
docs=db2.similarity_search(query)
docs[0].page_content
```

```python
# Similarity Search with score
docs=vectorDB.similarity_search_with_score(query)
docs
```

```python
# Retriever Option
retriever=vectorDB.as_retriever()
retriever.invoke(query)[0].page_content
```
