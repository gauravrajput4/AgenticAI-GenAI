### Pydantic Basics: Creating and using Models
* Pydantic Models are the foundation of data validation in Python, They use Python type annotations to deine the strucutre are validate data at runtime. Here's a detailed exploration of Basic model creation with several example.

```python
from tkinter.scrolledtext import example

from pydantic import BaseModel, ValidationError
```

```python
from dataclasses import dataclass

@dataclass
class Person():
    name: str
    age: int
    city: str

person=Person(name='Gaurav', age=22,city="Mohali")
print(person)
```

#### Model with Optional Fields
Add optional fields using Python's Optional type:
* Optional[type]:  Indicates the field can None
* Default value (=None or = True): Makes the field optional
* Required fields must still be provided
* Pydantic validates types even for optional fields when values are provided

```python
from typing import Optional

class Employee(BaseModel):
    id: int
    name: str
    department:str
    salary: Optional[float] = None  # Optional field with default value None
    is_active: Optional[bool] = True  # Optional field with default value True
```

```python
emp1=Employee(id=1, name='Alice', department='HR')
print(emp1)
```

```python
emp2=Employee(id=1, name='Alice', department='IT',salary=60000, is_active=False)
print(emp2)
```

```python
from typing import List

class ClassRoom(BaseModel):
    room_number: int
    students: List[str]  # List of student names
    capacity:int
```

```python
classroom=ClassRoom(room_number=101, students=['Alice', 'Bob', 'Charlie'], capacity=30)
print(classroom)
```

```python
try:
    invalid_val=ClassRoom(room_number=102, students=['Alice', 124], capacity=30)
    print(invalid_val)
except ValidationError as e:
    print(e)
```

#### Model with Nested Models
Create complex structure with nested Models:


```python
from pydantic import BaseModel

class Address(BaseModel):
    street: str
    city: str
    zipcode: str

class Customer(BaseModel):
    name: str
    email: str
    address: Address  # Nested model

# Create a customer with nested address
customer=Customer(
    name='Akhil',
    email='akhilrajput40@gmail.com',
    address=Address(
        street='Street 1',
        city='City 1',
        zipcode='123456',
    ))
print(customer)
```

#### Pydantic Fields: customization and Constraints
The Fields function is Pydantic enhance more fields beyond basic type hints by allowing you to specify validation rules, default valuem aliases and more. Here's a comprehensive tutorial with examples.

```python
from pydantic import BaseModel, Field

class Item(BaseModel):

    name: str=Field(min_length=3,max_length=50)
    price:float=Field(gt=0,le=1000)
    quantity:int=Field(ge=0)
```

```python
item1=Item(name='Laptop', price=999.99, quantity=10)
print(item1)
```

```python
    class User(BaseModel):
    username: str = Field(..., min_length=3, max_length=20,description="Uniques username for the user")
    email: str = Field(default_factory=lambda : "user@example.com",description="Default email is user@example.com")  # Simple email validation
    age: Optional[int] = Field(ge=0,description="User age default age 18",default=18)  # Optional field with minimum value constraint

## example 1
user1=User(
    username="Alice",
    email="alice@example,con=m",
    age=18,
)
print(user1)

# example 2
user2 = User(
    username="Bob",
    email="bob@example.com",
    age=-1,
)
```
