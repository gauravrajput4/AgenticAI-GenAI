##### In this quickstart we'll see how to:
* Get setup with Langchain, Langsmith, anf LangServe
* Use the most basic and common components of langchain: Prompt templates, models, and output parsers.
* Build a simple application with Langchain
* Trace you Application with LangSmith
* Serve you Application with LangServe

```python
import os

from dotenv import load_dotenv
from langsmith import utils

# Clear cached LangSmith environment variables
utils.get_env_var.cache_clear()

# Reload .env and replace old cached values
load_dotenv(override=True)

required_variables = [
    "LANGSMITH_API_KEY",
    "LANGSMITH_PROJECT",
]

for variable in required_variables:
    if not os.getenv(variable):
        raise ValueError(f"{variable} is missing from .env")

print("Tracing enabled:", os.getenv("LANGSMITH_TRACING_V2"))
print("Project:", os.getenv("LANGSMITH_PROJECT"))
print("Endpoint:", os.getenv("LANGSMITH_ENDPOINT"))
print("API key loaded:", bool(os.getenv("LANGSMITH_API_KEY")))
print("Workspace ID loaded:", bool(os.getenv("LANGSMITH_WORKSPACE_ID")))
```

```python
from langchain_ollama import ChatOllama

llm = ChatOllama(
    model="gemma2:2b",
    temperature=0
)
print(llm)
```

```python
response = llm.invoke("What is Generative AI?")

print(response.content)
```

```python
### Chatprompt Template
from langchain_core.prompts import ChatPromptTemplate
prompt=ChatPromptTemplate.from_messages(
    [
        ("system","You are an expert AI Engineer. Provide me answer based on the question"),
        ("user","{input}")
    ]
)
prompt
```

```python
chain=prompt|llm
response=chain.invoke({"input":"Can you tell me about langSmith"})
print(response.content)
```

```python
### String output parser
from langchain_core.output_parsers import StrOutputParser
output_parser=StrOutputParser()
chain=prompt|llm|output_parser
```

```python
response=chain.invoke({"input":"Can you tell me about langSmith"})
print(response)
```
